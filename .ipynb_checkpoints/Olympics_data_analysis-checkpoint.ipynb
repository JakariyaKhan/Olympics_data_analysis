```python
import numpy as np
import pandas as pd
from matplotlib import pyplot as plt
import seaborn as sns
import streamlit as st
```

```python
@st.cache_data
def load_data():
    games = pd.read_csv(
        "athlete_events.csv",
        low_memory=False
    )

    regions = pd.read_csv(
        "noc_regions.csv"
    )

    return games, regions


games, regions = load_data()
```

## Joining both dataframes

```python
df = games.merge(
    regions,
    on="NOC",
    how="left"
)
```

```python
df.sample(10)
```

## Medal_Analysis

```python
medal_count = df['Medal'].value_counts()

print(medal_count)
```

```python
medals=df.groupby("Medal")
medals.count()
```

```python
gold_medalist=df[df["Medal"]=="Gold"]
```

```python
gold_medalist.head()
```

```python
gold_medalist.isnull().any()
```

```python
gold_medalist=gold_medalist[np.isfinite(gold_medalist['Age'])]
```

```python
plt.figure(figsize=(20, 10))

sns.countplot(
    x='Age',
    data=gold_medalist,
    hue='Age',
    palette='viridis',
    legend=False
)

plt.title('Age Distribution of Gold Medalists')
plt.xlabel('Age')
plt.ylabel('Number of Gold Medalists')

plt.tight_layout()
plt.show()
```

It seems that there are people above Age of 50 winning Gold Medal and Below the Age of 20 as well.

```python
gold_medalist['ID'][gold_medalist['Age'] > 50].count()
```

```python
gold_medalist['ID'][gold_medalist['Age'] < 20].count()
```

```python
senior_medalist=gold_medalist['Sport'][gold_medalist['Age']>50]
```

```python
plt.figure(figsize=(20, 10))

sns.countplot(
    x=senior_medalist,
    hue=senior_medalist,
    palette='viridis',
    legend=False
)

plt.title('Gold Medals for Athletes Over 50')
plt.xlabel('Sport')
plt.ylabel('Number of Gold Medals')



plt.tight_layout()
plt.show()
```

```python
young_medalist=gold_medalist['Sport'][gold_medalist['Age']<=20]
```

```python
plt.figure(figsize=(12, 8))

top_10_young_sports = young_medalist.value_counts().head(10)

sns.barplot(
    x=top_10_young_sports.values,
    y=top_10_young_sports.index
)

plt.title('Top 10 Sports with Most Gold Medals Won by Athletes Under 20')
plt.xlabel('Number of Gold Medals')
plt.ylabel('Sport')

plt.tight_layout()
plt.show()
```
